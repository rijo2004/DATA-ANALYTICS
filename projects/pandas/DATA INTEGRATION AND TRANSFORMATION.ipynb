```python
#basic merging

import pandas as pd
df1 = pd.DataFrame({
    "id":[1,2,3],
    "name":["asha","binu","cathy"]
})
df2 = pd.DataFrame({
    "id":[1,2,4],
    "mark":[85,90,75]
})

merged_df = pd.merge(df1,df2,on="id") 
merged_df
```

#TYPES OF MERGING(INNER,RIGHT,LEFT,OUTER)

```python
#INNER MERGING
pd.merge(df1,df2,on="id",how = "inner")
```

```python
#LEFT MERGE
pd.merge(df1,df2,on ="id",how = "left")
```

```python
#RIGHT MERGE
pd.merge(df1,df2,on="id",how= "right")
```

```python
#OUTER MERGE
pd.merge(df1,df2,on= "id",how="outer")
```

#MERGE ON DIFFERNCE COLUMN NAMES

```python
import pandas as pd
df1 = pd.DataFrame({
    "id":[1,2,3],
    "name":["asha","binu","cathy"]
})
df2 = pd.DataFrame({
    "student_id":[1,2,4],
    "mark":[85,90,75]
})
merged_df=pd.merge(df1,df2,left_on="id",right_on="student_id")
merged_df
```

```python
#CONCATENATION (ROW WISE)
pd.concat([df1,df2],axis = 0)
```

```python
pd.concat([df1,df2],axis = 1)
```

```python
#APPLY
import pandas as pd
mark = pd.Series([45,60,70,80])
def grade(x):
   if x>50:
       return  "pass"
   else:
      return"fail"
mark.apply(grade)
```
