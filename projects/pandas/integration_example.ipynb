```python
import pandas as pd
df1=pd.DataFrame({
    "id":[1,2,3,4],
    "course":["ui/ux","data analyst","phython","c"],
    "Name":["rijo","sanju","aflah","vishnu"],
})
df2=pd.DataFrame({
    "id":[1,2,3,5],
    "mark":[20,40,80,60]
})
merged_df = pd.merge(df1,df2,on="id")
merged_df
```

```python
pd.merge(df1,df2,on="id",how="inner")
```

```python
pd.merge(df1,df2,on="id",how="left")
```

```python
pd.merge(df1,df2,on="id",how="right")
```

```python
pd.merge(df1,df2,on="id",how="outer")
```

```python
import pandas as pd
df1=pd.DataFrame({
    "student_id":[1,2,3,4],
    "course":["ui/ux","data analyst","phython","c"],
    "Name":["rijo","sanju","aflah","vishnu"],
})
df2=pd.DataFrame({
    "id":[1,2,3,5],
    "mark":[20,40,80,60]
})
merged_df = pd.merge(df1,df2,left_on="student_id",right_on="id")
merged_df
```

```python
pd.merge(df1,df2,left_on="student_id",right_on="id",how="inner")
```

```python
pd.merge(df1,df2,left_on="student_id",right_on="id",how="left")
```

```python
pd.merge(df1,df2,left_on="student_id",right_on="id",how="right")
```

```python
pd.merge(df1,df2,left_on="student_id",right_on="id",how="outer")
```
