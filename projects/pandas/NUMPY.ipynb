#agregation functions

```python
import numpy as np
ay = np.array([2,4,6])
print(np.sum(ay))
print(np.median(ay))
print(np.mean(ay))
print(np.std(ay))
print(np.var(ay))
print(np.min(ay))
print(np.max(ay))
print(np.argmax(ay))
print(np.argmin(ay))
```
