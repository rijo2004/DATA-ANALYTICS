```python
import pandas as pd

data ={
    "name":["asha","rahul","anu"],
    "age":[20,21,19],
    "city":["kochi","delhi","chennai"]
}
df = pd.DataFrame(data)
df
```

```python
#selecting by label(loc)
df.loc[1]
```

```python
df.loc[0:1,["name","age"]]
```

```python
#selecting by position(intex)(iloc)
df.iloc[1]
```

```python
df.iloc[0:2,0:2]
```

```python
#DIFFERENCE IN SLICEING(ending included)
df.loc[0:1]
```

```python
#(ending excluded)
df.iloc[0:1]
```

```python
#SELECTING SINGLE VALUE
df.loc[2,"city"]
```

```python
df.iloc[2,2]
```
