# longest word

```python
#EXAMPLE 1
s = input("enter a sentence:")
longest = " "

for i in s.split():
    if len(i) > len(longest):
        longest = i
        
print("the longest word is:",longest)
```

# Shift All Zeros to the End

```python
x =[0,1,27,0,46,54]
y =[ ]
for i in x:
    if i != 0:
        y.append(i)
        
    for i in x:
          if i == 0:
              y.append(i)
        
print(y)
```
