```python
#50 to 70 avg
#20 or below faild
#20 to 50 pass
#70 to 80 good
# 80 to 90 excellent
#above 90 outstanding

score = int(input("enter ur mark: "))

if score >= 90 and score<=100:
    print(score,"OUTSTANDING")
elif score >= 80 and score<=89:
    print(score,"EXCELLENT")
elif score >= 70 and score<=79:
    print(score,"GOOD")
elif score >= 50 and score<=59:
    print(score,"AVERAGE")
elif score >= 20 and score<=29:
    print(score,"PASS")
elif score>=0 and score<=19:
    print(score,"you are failed")
else:
    print("invalid number")
    
```
