```python
#creating file
with open("handling.txt","w") as file:
    file.write("successfully created")
```

```python
#reading file
with open("handling.txt","r") as file:
     print(file.read())
```

```python
#overwrite file
with open("handling.txt","w") as file:
     file.write("this folder is for phython handling")
```

```python
#adding new line in the file
with open("handling.txt","a") as file:
    file.write("\nthis is a new line for this file")
```

```python
#reading full content
with open("handling.txt","r") as file:
   print(file.read())
```

```python
# read line by line
with open("handling.txt","r")as file:
    for line in file:
        print(line.strip())
```

```python
# read only one line
with open("handling.txt","r") as file:
    print(file.readline())
```

```python
#working with file pointer
with open("handling.txt","r") as file:
    print(file.read(5))
    file.seek(0)
    print(file.read())
```

```python
#Tell current position
with open("handling.txt","r") as file:
    print(file.tell()) #before positioning
    file.read(10)
    print(file.tell()) #after positioning
    
```

```python
#changing the content into a list
with open("handling.txt","r") as file:
    lines = file.readlines()
    print(lines)
```

```python
#writing multiple lines 
lines = ["my name is rijo\n","I am from pariyapuram\n","i am a data analytics intern at softronics\n"]
with open("handling.txt","w") as file:
    file.writelines(lines)
```

```python
#copy file
with open("handling.txt","r") as src:
    content = src.read()
    
with open("copy.txt","w") as dest:
     dest.write(content)
    
```

```python
#rename file
import os
os.rename("handling.txt","newhandling.txt")
```

```python
# remove file
import os
os.remove("copy.txt")
```

```python
#csv file creating
import csv

with open("data.csv","w",newline="")as file:
    writer = csv.writer(file)
    writer.writerow(["NAME","AGE"])
    writer.writerow(["rijo","21"])
```

```python
#json file creating
import json
data = {"name":"rijo","age":21}
with open("student.json","w") as file:
    json.dump(data,file)
with open("student.json","r") as file:
    print(json.load(file))
```
