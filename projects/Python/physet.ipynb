```python
A = {"apple","banana","cherry"}
B = {"orange","grape","apple"}
c = A.union(B)
print("union of set :",c)
```

```python
A = {"apple","banana","cherry"}
B = {"orange","grape","apple"}
c = A.intersection(B)
print("intersection of set :",c)
```

```python
A = {"apple","banana","cherry"}
B = {"orange","grape","apple"}
c = A.difference(B)
print("intersection of set :",c)
```

```python
A = {"apple","banana","cherry"}
B = {"orange","grape","apple"}
c = A.symmetric_difference(B)
print("intersection of set :",c)
```

```python
A = {"apple","banana","cherry"}
B = {"orange","grape","apple"}
c = A.isdisjoint(B)
print(c)
```

```python
A = {"apple","orange","grape","banana","cherry"}
B = {"orange","grape","apple"}
c = A.issuperset(B)
print(c)
```

```python
A = {"apple","orange","grape","banana","cherry"}
B = {"orange","banana","apple"}
c = B.issubset(A)
print(c)
```
